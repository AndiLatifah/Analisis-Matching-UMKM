### mengubah X dan Z menjadi Unmatch dan Match

```python
import pandas as pd

df = pd.read_csv("SudahLabel-Tifa(terpisahJW)part3.csv")
print("Jumlah data:", len(df))
df.head(10)
```

```python

df["label"] = df["label"].astype(str).str.strip().str.upper()

df["label"] = df["label"].replace({
    "X": "unmatch",
    "Z": "match"
})

df["label"] = df["label"].str.lower()

df.to_csv("SudahLabel-Tifa(terpisahJW)part33.csv", index=False)

print("File berhasil disimpan.")
```
