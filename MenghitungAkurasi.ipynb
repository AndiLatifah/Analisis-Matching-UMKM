```python
import pandas as pd

df = pd.read_csv("DataAKurasi_PartRappocini.csv")  # ganti nama file kamu

# Ubah label manual
df["y_true"] = df["label"].map({"Z":1, "X":0})

results = []

# =====================
# Uji Jaro-Winkler
# =====================
for t in range(85, 101):  # threshold 85–100
    y_pred = (df["score_jw"] >= t).astype(int)
    accuracy = (y_pred == df["y_true"]).mean()
    results.append(["JW", t, accuracy])

# =====================
# Uji Levenshtein
# =====================
for t in range(75, 101):  # threshold 75–100
    y_pred = (df["score_lev"] >= t).astype(int)
    accuracy = (y_pred == df["y_true"]).mean()
    results.append(["LEV", t, accuracy])

result_df = pd.DataFrame(results, columns=["Algoritma","Threshold","Accuracy"])

# Tampilkan 5 terbaik tiap algoritma
print("\nTop 5 JW:")
print(result_df[result_df["Algoritma"]=="JW"].sort_values("Accuracy", ascending=False).head(6))

print("\nTop 5 Levenshtein:")
print(result_df[result_df["Algoritma"]=="LEV"].sort_values("Accuracy", ascending=False).head(6))
```

```python
import pandas as pd

df = pd.read_csv("DataAKurasi_PartRappocini.csv")

# ubah label manual
df["y_true"] = df["label"].map({"Z":1, "X":0})

threshold = 0.97  # ganti sesuai yang mau kamu cek

# Prediksi JW
df["y_pred_jw"] = (df["score_jw"] >= threshold).astype(int)

# Hitung confusion matrix manual
TP = ((df["y_true"] == 1) & (df["y_pred_jw"] == 1)).sum()
TN = ((df["y_true"] == 0) & (df["y_pred_jw"] == 0)).sum()
FP = ((df["y_true"] == 0) & (df["y_pred_jw"] == 1)).sum()
FN = ((df["y_true"] == 1) & (df["y_pred_jw"] == 0)).sum()

accuracy = (TP + TN) / len(df)

print("Threshold:", threshold)
print("TP:", TP)
print("TN:", TN)
print("FP:", FP)
print("FN:", FN)
print("Accuracy:", round(accuracy,4))
```

```python
import pandas as pd

# =========================
# Load & rapikan data
# =========================
df = pd.read_csv("DataAKurasi_PartRappocini.csv")

df.columns = df.columns.str.strip()
df["label"] = df["label"].astype(str).str.strip().str.upper()

df["score_jw"]  = pd.to_numeric(df["score_jw"], errors="coerce")
df["score_lev"] = pd.to_numeric(df["score_lev"], errors="coerce")

# Ground truth: Z=1 (match), X=0 (tidak match)
df["y_true"] = df["label"].map({"Z": 1, "X": 0})

# Buang baris yang labelnya tidak valid atau skor kosong
df_eval = df.dropna(subset=["y_true", "score_jw", "score_lev"]).copy()
df_eval["y_true"] = df_eval["y_true"].astype(int)

# =========================
# Fungsi hitung TP/TN/FP/FN + Accuracy
# =========================
def confusion_and_acc(y_true, y_pred):
    tp = ((y_true == 1) & (y_pred == 1)).sum()
    tn = ((y_true == 0) & (y_pred == 0)).sum()
    fp = ((y_true == 0) & (y_pred == 1)).sum()
    fn = ((y_true == 1) & (y_pred == 0)).sum()
    acc = (tp + tn) / len(y_true) if len(y_true) else 0
    return tp, tn, fp, fn, acc

results = []
y_true = df_eval["y_true"]

# =========================
# Evaluasi JW
# =========================
for t_int in range(85, 100):   # 0.85 .. 0.99
    t = t_int / 100
    y_pred = (df_eval["score_jw"] >= t).astype(int)
    tp, tn, fp, fn, acc = confusion_and_acc(y_true, y_pred)
    results.append(["JW", t, tp, tn, fp, fn, acc])

# =========================
# Evaluasi Levenshtein
# =========================
for t_int in range(75, 100):   # 0.75 .. 0.99
    t = t_int / 100
    y_pred = (df_eval["score_lev"] >= t).astype(int)
    tp, tn, fp, fn, acc = confusion_and_acc(y_true, y_pred)
    results.append(["LEV", t, tp, tn, fp, fn, acc])

result_df = pd.DataFrame(
    results, columns=["Algoritma", "Threshold", "TP", "TN", "FP", "FN", "Accuracy"]
)

# =========================
# Tampilan lebih menarik
# =========================
pd.set_option("display.width", 160)
pd.set_option("display.max_columns", 20)

def show_top(algo, n=10):
    sub = result_df[result_df["Algoritma"] == algo].sort_values(
        ["Accuracy", "TP"], ascending=[False, False]
    ).head(n).copy()

    # format persen biar enak
    sub["Accuracy(%)"] = (sub["Accuracy"] * 100).round(3)
    sub = sub.drop(columns=["Accuracy"])

    print(f"\n{'='*60}")
    print(f"TOP {n} {algo} (urut Accuracy tertinggi, lalu TP tertinggi)")
    print(f"{'='*60}")
    print(sub.to_string(index=False))

def show_best(algo):
    best = result_df[result_df["Algoritma"] == algo].sort_values(
        ["Accuracy", "TP"], ascending=[False, False]
    ).head(1).copy()

    best["Accuracy(%)"] = (best["Accuracy"] * 100).round(3)
    print(f"\n>>> BEST {algo}")
    print(best[["Algoritma", "Threshold", "TP", "TN", "FP", "FN", "Accuracy(%)"]].to_string(index=False))

# ringkasan data label
print("\nRINGKASAN LABEL (setelah dibersihkan):")
print(df_eval["y_true"].value_counts().rename(index={1:"MATCH(Z)", 0:"TIDAK(X)"}).to_string())

# tampilkan top
show_top("JW", n=10)
show_top("LEV", n=10)

# tampilkan best
show_best("JW")
show_best("LEV")

# (opsional) simpan hasil evaluasi lengkap
result_df.to_csv("EVALUASI_AKURASI_JW_LEV_dengan_CM.csv", index=False)
```

```python
import pandas as pd

df = pd.read_csv("SudahLabel-Linda.csv")

df.columns = df.columns.str.strip()
df["label"] = df["label"].astype(str).str.strip().str.upper()

df["score_jw"]  = pd.to_numeric(df["score_jw"], errors="coerce")
df["score_lev"] = pd.to_numeric(df["score_lev"], errors="coerce")

# Ground truth: Z=1 (match), X=0 (tidak match)
df["y_true"] = df["label"].map({"Z": 1, "X": 0})

# Buang baris yang labelnya tidak valid atau skor kosong
df_eval = df.dropna(subset=["y_true", "score_jw", "score_lev"]).copy()
df_eval["y_true"] = df_eval["y_true"].astype(int)

# =========================
# hitung TP/TN/FP/FN + Accuracy
def confusion_and_acc(y_true, y_pred):
    tp = ((y_true == 1) & (y_pred == 1)).sum()
    tn = ((y_true == 0) & (y_pred == 0)).sum()
    fp = ((y_true == 0) & (y_pred == 1)).sum()
    fn = ((y_true == 1) & (y_pred == 0)).sum()
    acc = (tp + tn) / len(y_true) if len(y_true) else 0
    return tp, tn, fp, fn, acc

results = []
y_true = df_eval["y_true"]

# =========================
# Evaluasi JW
for t_int in range(85, 100):   # 0.85 .. 0.99
    t = t_int / 100
    y_pred = (df_eval["score_jw"] >= t).astype(int)
    tp, tn, fp, fn, acc = confusion_and_acc(y_true, y_pred)
    results.append(["JW", t, tp, tn, fp, fn, acc])

# =========================
# Evaluasi Levenshtein
for t_int in range(75, 100):   # 0.75 .. 0.99
    t = t_int / 100
    y_pred = (df_eval["score_lev"] >= t).astype(int)
    tp, tn, fp, fn, acc = confusion_and_acc(y_true, y_pred)
    results.append(["LEV", t, tp, tn, fp, fn, acc])

result_df = pd.DataFrame(
    results, columns=["Algoritma", "Threshold", "TP", "TN", "FP", "FN", "Accuracy"]
)

pd.set_option("display.width", 160)
pd.set_option("display.max_columns", 20)

def show_top(algo, n=10):
    sub = result_df[result_df["Algoritma"] == algo].sort_values(
        ["Accuracy", "TP"], ascending=[False, False]
    ).head(n).copy()

    # format persen biar enak
    sub["Accuracy(%)"] = (sub["Accuracy"] * 100).round(3)
    sub = sub.drop(columns=["Accuracy"])

    print(f"\n{'='*60}")
    print(f"TOP {n} {algo} (urut Accuracy tertinggi, lalu TP tertinggi)")
    print(f"{'='*60}")
    print(sub.to_string(index=False))

def show_best(algo):
    best = result_df[result_df["Algoritma"] == algo].sort_values(
        ["Accuracy", "TP"], ascending=[False, False]
    ).head(1).copy()

    best["Accuracy(%)"] = (best["Accuracy"] * 100).round(3)
    print(f"\n>>> BEST {algo}")
    print(best[["Algoritma", "Threshold", "TP", "TN", "FP", "FN", "Accuracy(%)"]].to_string(index=False))

# ringkasan data label
print("\nRINGKASAN LABEL (setelah dibersihkan):")
print(df_eval["y_true"].value_counts().rename(index={1:"MATCH(Z)", 0:"TIDAK(X)"}).to_string())

# tampilkan top
show_top("JW", n=10)
show_top("LEV", n=10)

# tampilkan best
show_best("JW")
show_best("LEV")

# (opsional) simpan hasil evaluasi lengkap
result_df.to_csv("EVALUASI_AKURASI_JW_LEV(SudahLabel-Linda).csv", index=False)
```
