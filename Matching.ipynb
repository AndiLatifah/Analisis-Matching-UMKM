```python
import pandas as pd
from rapidfuzz import process, distance

BPS_CSV  = "BPS_Rappocini_clean.csv"
MAPS_CSV = "Scraping_Rappocini_clean.csv"

OUT_JW_CSV  = "MATCHING_Rappocini_JW.csv"
OUT_LEV_CSV = "MATCHING_Rappocini_LEV.csv"

df_bps  = pd.read_csv(BPS_CSV)
df_maps = pd.read_csv(MAPS_CSV)

# Pastikan tidak ada NaN
df_bps["Nama Usaha"] = df_bps["Nama Usaha"].fillna("").astype(str)
df_maps["title"]     = df_maps["title"].fillna("").astype(str)

# List nama usaha BPS
bps_names = df_bps["Nama Usaha"].tolist()

results_jw = []
results_lev = []

# PROSES MATCHING
for _, row in df_maps.iterrows():
    query_name = row["title"]

    if query_name.strip() == "":
        continue

    # MATCHING JARO-WINKLER
    best_jw = process.extractOne(
        query_name,
        bps_names,
        scorer=distance.JaroWinkler.normalized_similarity
    )

    nama_bps_jw, score_jw, idx_jw = best_jw
    bps_row_jw = df_bps.iloc[idx_jw]

    results_jw.append({
        "input_id": row.get("input_id", ""),
        "title_maps": query_name,
        "address_maps": row.get("address", ""),

        "IDSBR_best": bps_row_jw.get("IDSBR", ""),
        "nama_bps_best": nama_bps_jw,
        "alamat_bps_best": bps_row_jw.get("Alamat", ""),

        "score_jw": round(score_jw, 42)
    })

    # MATCHING LEVENSHTEIN
    best_lev = process.extractOne(
        query_name,
        bps_names,
        scorer=distance.Levenshtein.normalized_similarity
    )

    nama_bps_lev, score_lev, idx_lev = best_lev
    bps_row_lev = df_bps.iloc[idx_lev]

    results_lev.append({
        "input_id": row.get("input_id", ""),
        "title_maps": query_name,
        "address_maps": row.get("address", ""),

        "IDSBR_best": bps_row_lev.get("IDSBR", ""),
        "nama_bps_best": nama_bps_lev,
        "alamat_bps_best": bps_row_lev.get("Alamat", ""),

        "score_lev": round(score_lev, 2)
    })

df_result_jw = pd.DataFrame(results_jw)
df_result_lev = pd.DataFrame(results_lev)

df_result_jw.to_csv(OUT_JW_CSV, index=False)
df_result_lev.to_csv(OUT_LEV_CSV, index=False)

print("Selesai!")
print("File Jaro-Winkler :", OUT_JW_CSV)
print("File Levenshtein  :", OUT_LEV_CSV)

print("\nContoh hasil Jaro-Winkler:")
print(df_result_jw.head(10))

print("\nContoh hasil Levenshtein:")
print(df_result_lev.head(10))
```
