```python
from rapidfuzz import process, distance
import pandas as pd

df_maps = pd.read_csv("BPS_Rappocini_clean.csv")


choices = df_maps["Nama Usaha"].astype(str)

queries = [
    "aicomp.",
    "apotik aldika farma",
    "dhiba cell",
    "tala yamaha motor",
    "bersama bangunan",
    "cv. alief karya tehnik"
]

# menampilkan Top-5 kandidat
def tampilkan_hasil(query):

    print("="*80)
    print("QUERY :", query)

    print("\nJARO-WINKLER")
    jw = process.extract(
        query,
        choices,
        scorer=distance.JaroWinkler.normalized_similarity,
        limit=5
    )

    for nama, score, idx in jw:
        alamat = df_maps.loc[idx, "Alamat"]
        print(f"{score:.3f}")
        print("Nama   :", nama)
        print("Alamat :", alamat)
        print()

    print("\nLEVENSHTEIN")
    lev = process.extract(
        query,
        choices,
        scorer=distance.Levenshtein.normalized_similarity,
        limit=5
    )

    for nama, score, idx in lev:
        alamat = df_maps.loc[idx, "Alamat"]
        print(f"{score:.3f}")
        print("Nama   :", nama)
        print("Alamat :", alamat)
        print()

for q in queries:
    tampilkan_hasil(q)
```
